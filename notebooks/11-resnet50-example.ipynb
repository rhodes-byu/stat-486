```python
import matplotlib.pyplot as plt
import os
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input, decode_predictions
from tensorflow.keras.preprocessing import image
import numpy as np
```

```python
# Find a few images and place them in a directory called 'myimages'
# images should be .jpg, .jpeg, .JPG, .JPEG, .png, .PNG
image_dir = 'myimages/'
extensions = ['jpg','jpeg','JPG','JPEG','png','PNG']
images = [file for file in os.listdir(image_dir) if file.split('.')[-1] in extensions]
images.sort() # sort the images in alphabetical order
```

```python
len(images)
```

```python
# These dimensions should match the number of images you have
n_cols = 2
n_rows = 1
```

```python
## (I cropped the images in photoshop to be square)
plt.figure(figsize=(25,25))
for row in range(n_rows):
    for col in range(n_cols):
        index = n_cols * row + col
        img_path = image_dir + images[index]
        img = image.load_img(img_path, target_size=(224, 224))
        plt.subplot(n_rows, n_cols, index + 1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(images[index])
#plt.subplots_adjust(wspace=0.2, hspace=0.5)
plt.show()
```

```python
top_preds = []
ntop = 3
# Load the ResNet50 model with pretrained weights
model = ResNet50(weights='imagenet')

# Define the folder containing the images
folder_path = image_dir

# Loop through the images in the folder
for filename in images:
    # Load the image and resize it to fit the ResNet50 input shape
    img = image.load_img(os.path.join(folder_path, filename), target_size=(224, 224))

    # Convert the image to an array
    x = image.img_to_array(img)

    # Add a dimension to the array to make it compatible with ResNet50 input shape
    x = np.expand_dims(x, axis=0)

    # Preprocess the input image according to ResNet50 requirements
    x = preprocess_input(x)

    # Use the model to classify the image
    preds = model.predict(x)

    # Decode the predictions and print the top 3 predicted classes
    decoded_preds = decode_predictions(preds, top=ntop)[0]
    top_preds.append(decoded_preds)

#     print('Image:', filename)
#     for pred in decoded_preds:
#         print(pred[1], pred[2])
```

```python
plt.figure(figsize=(25,25))
for row in range(n_rows):
    for col in range(n_cols):
        index = n_cols * row + col
        img_path = image_dir + images[index]
        img = image.load_img(img_path, target_size=(224, 224))
        plt.subplot(n_rows, n_cols, index + 1)
        plt.imshow(img)
        plt.axis('off')
        plt.title(images[index])

        #table_data = [['Class', 'Probability']]
        table_data = []
        for pred in top_preds[index]:
            table_data.append([pred[1], f'{pred[2]*100:.2f}%'])
        table = plt.table(cellText=table_data, loc='bottom', cellLoc='center', colWidths=[0.9, 0.4], cellColours=[['#e1f5fe']*2] * ntop,  edges='closed')
        table.auto_set_font_size(False)
        table.set_fontsize(10)
        table.scale(1, 2)
plt.subplots_adjust(wspace=0.2, hspace=.6)
plt.show()
```
